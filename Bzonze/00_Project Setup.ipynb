```python
spark.sql("""
    CREATE CATALOG IF NOT EXISTS globalretail
""")

print("Catalog created or already exists")
```

```python
schemas=["bronze", "silver", "gold", "validation"]

for schema in schemas:
  spark.sql(f"""
    CREATE SCHEMA IF NOT EXISTS globalretail.{schema}
  """)
  
  print(f"Schema {schema} created ")
```

```python
spark.sql("""
          SHOW SCHEMAS IN globalretail
          """).show(truncate=False)
```
