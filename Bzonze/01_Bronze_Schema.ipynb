```python
%sql

CREATE TABLE IF NOT EXISTS globalretail.bronze.customers_raw (
    customer_id       INT,
    first_name        STRING,
    last_name         STRING,
    email             STRING,
    phone             STRING,
    address_line1     STRING,
    city              STRING,
    state             STRING,
    country_code      STRING,
    postal_code       STRING,
    loyalty_tier      STRING,
    marketing_opt_in  STRING,
    source_system     STRING,
    updated_at        TIMESTAMP,
    is_active         STRING,
    ingested_at       TIMESTAMP
)
USING DELTA;
```

```python
%sql

CREATE TABLE IF NOT EXISTS globalretail.bronze.stores_raw(
    store_id       INT            ,
    store_name     STRING         ,
    city           STRING         ,
    country_code   STRING         ,
    timezone_name  STRING        ,
    region         STRING,
    opened_date    DATE          ,
    updated_at     TIMESTAMP      ,
    is_active      STRING        ,
    ingested_at    TIMESTAMP
)
USING DELTA;
```

```python
%sql

CREATE TABLE IF NOT EXISTS globalretail.bronze.products_raw (
    product_id      INT,
    sku             STRING,
    product_name    STRING,
    category        STRING,
    subcategory     STRING,
    brand           STRING,
    unit_price      DECIMAL(12,2),
    unit_cost       DECIMAL(12,2),
    currency_code   STRING,
    reorder_point   INT,
    is_active       STRING,
    updated_at      TIMESTAMP,
    ingested_at     TIMESTAMP
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS globalretail.bronze.sales_raw(
    order_id        STRING,
    order_line_id   DECIMAL(10,0),
    order_ts_utc    STRING,
    customer_id     DECIMAL(10,0),
    product_id      DECIMAL(10,0),
    store_id        DECIMAL(10,0),
    channel         STRING,
    quantity        DECIMAL(10,0),
    unit_price      DECIMAL(12,2),
    discount_amount DECIMAL(12,2),
    tax_amount      DECIMAL(12,2),
    currency_code   STRING,
    payment_status  STRING,
    updated_at      STRING,
    ingested_at     TIMESTAMP

) USING DELTA;
```
